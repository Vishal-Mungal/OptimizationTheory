# Finding Local Minima and Maxima of a Single-Variable Function

This notebook will guide you through finding the local minimum and maximum of a single-variable function using symbolic computation with the `sympy` library.

```python
import sympy
from sympy import symbols, diff, solve
```

## 1. Define the Function

First, we need to define our single-variable function. Let's use `x` as our symbolic variable.

```python
# Define the symbolic variable
x = symbols('x')

# Define the function (you can change this to any single-variable function)
f_x = x**3 - 6*x**2 + 9*x + 1

print(f"The function is: f(x) = {f_x}")
```

## 2. Find Critical Points (First Derivative Test)

Local extrema occur at critical points, where the first derivative of the function is equal to zero or undefined. We'll calculate the first derivative and then solve for `x` where it equals zero.

```python
# Calculate the first derivative
f_prime_x = diff(f_x, x)
print(f"The first derivative f'(x) = {f_prime_x}")

# Find critical points by setting the first derivative to zero
critical_points = solve(f_prime_x, x)
print(f"Critical points (where f'(x) = 0): {critical_points}")
```

## 3. Apply the Second Derivative Test

To classify each critical point as a local minimum or maximum, we use the second derivative test:

- If `f''(c) > 0`, then `c` is a local minimum.
- If `f''(c) < 0`, then `c` is a local maximum.
- If `f''(c) = 0`, the test is inconclusive (further analysis is needed, but for typical polynomial functions, it's often an inflection point).

First, let's calculate the second derivative.

```python
# Calculate the second derivative
f_double_prime_x = diff(f_prime_x, x)
print(f"The second derivative f''(x) = {f_double_prime_x}")

print("\nClassifying critical points:")
local_minima = []
local_maxima = []

for cp in critical_points:
    second_deriv_at_cp = f_double_prime_x.subs(x, cp)
    if second_deriv_at_cp > 0:
        y_value = f_x.subs(x, cp)
        local_minima.append({'x': cp, 'y': y_value})
        print(f"At x = {cp}: f''({cp}) = {second_deriv_at_cp} > 0 => Local Minimum at ({cp}, {y_value})")
    elif second_deriv_at_cp < 0:
        y_value = f_x.subs(x, cp)
        local_maxima.append({'x': cp, 'y': y_value})
        print(f"At x = {cp}: f''({cp}) = {second_deriv_at_cp} < 0 => Local Maximum at ({cp}, {y_value})")
    else:
        y_value = f_x.subs(x, cp)
        print(f"At x = {cp}: f''({cp}) = {second_deriv_at_cp} => Inconclusive (possibly an inflection point) at ({cp}, {y_value})")


print("\n--- Summary ---")
if local_maxima:
    print("Local Maxima:")
    for lm in local_maxima:
        print(f"  x = {lm['x']}, y = {lm['y']}")
else:
    print("No local maxima found.")

if local_minima:
    print("Local Minima:")
    for lm in local_minima:
        print(f"  x = {lm['x']}, y = {lm['y']}")
else:
    print("No local minima found.")
```

## 4. Visualize the Function and Extrema (Optional)

We can also plot the function and highlight the found extrema to visualize our results.

```python
import matplotlib.pyplot as plt
import numpy as np

# Convert sympy expression to a function for plotting
f_numpy = sympy.lambdify(x, f_x, 'numpy')

# Generate x values for plotting
x_vals = np.linspace(-1, 5, 400)
y_vals = f_numpy(x_vals)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label=f'f(x) = {f_x}', color='blue')

# Plot local maxima
for lm in local_maxima:
    plt.scatter(lm['x'], lm['y'], color='red', s=100, zorder=5, label='Local Maximum' if 'Local Maximum' not in plt.gca().get_legend_handles_labels()[1] else "")
    plt.annotate(f'Max ({lm['x']:.2f}, {lm['y']:.2f})', (lm['x'], lm['y']), textcoords="offset points", xytext=(5,15), ha='center', color='red')

# Plot local minima
for lm in local_minima:
    plt.scatter(lm['x'], lm['y'], color='green', s=100, zorder=5, label='Local Minimum' if 'Local Minimum' not in plt.gca().get_legend_handles_labels()[1] else "")
    plt.annotate(f'Min ({lm['x']:.2f}, {lm['y']:.2f})', (lm['x'], lm['y']), textcoords="offset points", xytext=(5,-25), ha='center', color='green')

plt.title('Function with Local Minima and Maxima')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.show()
```
