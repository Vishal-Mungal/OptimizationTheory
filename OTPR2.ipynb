```python
import sympy
from sympy import symbols, diff, solve, Matrix, I, re, im
```

```python
x, y = symbols('x y')
f = x**2 + y**2 - 4*x -6*y +13

print(f"Original function: f({x}, {y}) = {f}")

fx = diff(f, x)
fy = diff(f, y)
print(f"Partial derivatives: f_x({x}) = {fx}, f_y({y}) = {fy}")


critical_points = solve([fx, fy], [x, y])
print(f"Critical points: {critical_points}")

fxx = diff(fx, x)
fxy = diff(fx, y)
fyy = diff(fy, y)
print(f"Hessian matrix: H = {Matrix([[fxx, fxy], [fxy, fyy]])}")

H = Matrix([[fxx, fxy], [fxy, fyy]])

print(f"\nHessian Matrix:\n{H}")

```

```python
fxx = diff(fx, x)
fyy = diff(fy, y)
fxy = diff(fx, y)
print(f"\nSecond partial derivative fxx: {fxx}")
print(f"Second partial derivative fyy: {fyy}")
print(f"Second partial derivative fxy: {fxy}")
```

```python
H = Matrix([[fxx, fxy], [fxy, fyy]])
print(f"\nHessian Matrix:\n{H}")
```

```python
det_H = H.det()
print(f"Determinant of Hessian (det(H)): {det_H}")

if det_H > 0:
    if fxx > 0:
        print("The stationary point is a local minimum.")
    elif fxx < 0:
        print("The stationary point is a local maximum.")
else:
    print("The Hessian test is inconclusive (det(H) <= 0).")
```
